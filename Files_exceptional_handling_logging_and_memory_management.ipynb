7.  How can you handle multiple exceptions in Python ?

9.  What is the difference between multithreading and multiprocessing ?

    # Multithreading
      
      1:- Threads are lightweight units of a process that share the same memory space.

     2:- In Python, multithreading is often used for tasks that involve I/O-bound operations (like file reading, web requests, etc.).

     3:- Due to the Global Interpreter Lock (GIL) in CPython (the standard Python implementation), only one thread executes Python bytecode at a time.

     4:- It's good for concurrent tasks where performance depends on waiting (e.g., waiting for a file to download or a user to input something).

    # Multiprocessing
   
   1:-Processes are independent execution units with their own memory space.

   2:- Multiprocessing is used for CPU-bound operations, where tasks need a lot of computation (e.g., number crunching, image processing).

   3:- Since each process has its own Python interpreter and memory, it bypasses the GIL—allowing true parallelism.

   4:- It's more powerful for CPU-intensive tasks but also heavier (more memory and startup time).

11.  What is memory management in Python ?

    # Ans :-

# Automatic Memory Allocation
Python automatically allocates memory for variables and objects when they are created.

You don’t need to manually declare or free memory like in C or C++.

# Garbage Collection
Python uses a garbage collector to automatically free up memory that is no longer in use.

It mainly relies on reference counting, where each object keeps track of how many references point to it.

When the reference count drops to zero, the object is deleted.

Python also has a cyclic garbage collector to clean up objects that refer to each other in a cycle, which reference counting alone can't handle.

# Private Heap Space
All Python objects and data structures are stored in a private heap—a chunk of memory managed internally by the Python interpreter.

Developers don’t access this heap directly; it's managed by the Python memory manager.

# Memory Pools (via pymalloc)
Python uses a specialized memory allocator called pymalloc for small objects.

This improves performance by reducing the overhead of frequent memory operations.

# Dynamic Typing
Python’s variables are dynamically typed, so memory is allocated based on the actual value assigned, not a fixed type or size.

15.  How does Python's garbage collection system work ?

#  Reference Counting (Primary Mechanism)
Every Python object keeps track of how many references point to it.

When you assign an object to a variable, its reference count increases.

When references are deleted or go out of scope, the count decreases.

If the reference count drops to zero, the object is immediately destroyed and memory is released.

#  Garbage Collector for Cycles (Secondary Mechanism)
Reference counting can’t detect circular references (e.g., two objects referring to each other).

Python has a cyclic garbage collector to handle these cases.

It periodically searches for groups of objects that are unreachable but still reference each other, and then removes them.

# The gc Module
Python provides the gc module to:

Manually trigger garbage collection.

Monitor and debug memory usage.

Control thresholds for when the cyclic collector runs.

#  Summary of the Garbage Collection Process:
Track all objects and their reference counts.

Immediately delete objects with zero references.

Periodically scan for unreachable cycles.

Clean up memory automatically without needing manual intervention.

23.  What are the challenges associated with memory management in Python ?

# Memory Leaks (Unreachable but Referenced Objects)
Cause: Memory leaks in Python occur when objects are no longer needed but are still being referenced, preventing the garbage collector from releasing them.

Example: Circular references (e.g., two objects referring to each other) can prevent the garbage collector from cleaning up memory, even though they are no longer accessible.

Solution: Python's garbage collector can detect some circular references, but developers need to be careful when dealing with complex object graphs. Tools like the gc module help manage this.

# Overhead of Garbage Collection
Cause: The garbage collector can introduce overhead in programs with high memory use, particularly in situations where objects are created and destroyed frequently.

Impact: Automatic garbage collection can sometimes slow down the application, especially if there are many objects to clean up.

Solution: In cases where performance is critical, you can use the gc module to control or disable garbage collection, though this requires careful management to avoid memory issues.

# Fragmentation
Cause: Memory fragmentation occurs when there are many small unused blocks of memory scattered throughout the memory pool, causing inefficient memory use.

Impact: Although Python's memory manager is designed to avoid fragmentation, long-running programs can still experience fragmentation, especially if objects are allocated and deallocated in unpredictable patterns.

Solution: Python uses a private heap for memory allocation, which helps reduce fragmentation, but developers should still be mindful of object management, especially for large programs.

# Object Overhead
Cause: Python objects have a certain amount of overhead for storing metadata, such as reference counts and other internal information. This can make the memory usage of objects higher than in lower-level languages like C or C++.

Impact: This object overhead can accumulate, particularly when you create large numbers of small objects.

Solution: To mitigate this, use built-in data structures (like tuple instead of list for small, fixed collections) or more efficient structures when appropriate. Avoid unnecessary object creation and try to reuse objects when possible.

# Memory Fragmentation with Large Objects
Cause: Large objects like big lists or large data structures can cause memory fragmentation, particularly when they grow and shrink in unpredictable ways.

Impact: If large objects are repeatedly resized, it can lead to inefficient memory use or even performance degradation.

Solution: In scenarios where large objects are needed, numpy arrays (for numerical data) or memoryviews are often more memory-efficient than regular Python lists or arrays.

25.  Why is it important to use multithreading in certain applications ?

#  Improved Application Performance
Concurrency: Multithreading allows the program to perform multiple tasks at the same time. This is particularly useful for I/O-bound tasks (e.g., reading/writing files, network requests, database queries) where the program spends time waiting for data to be read or sent.

Better CPU Utilization: While Python’s Global Interpreter Lock (GIL) limits true parallel execution in some cases, multithreading can still be useful when dealing with I/O-bound tasks. By having threads wait for I/O operations concurrently, the CPU can be kept busy with other threads, improving overall responsiveness.

# Non-blocking Operations
In a single-threaded program, if one task is waiting for I/O, the entire program might be blocked, unable to perform any other task.

With multithreading, while one thread is waiting (for example, for a response from a remote server), other threads can continue executing, making the application more responsive and capable of handling multiple operations at once without stalling.

# Real-time Applications
Certain applications, like video games, live streaming services, or real-time data processing systems, need to handle multiple tasks in parallel (like handling user input, rendering graphics, and downloading data) without delay. Multithreading allows the application to process these tasks concurrently in the background.

#  Improved User Experience (UI responsiveness)
In GUI-based applications, like those built with frameworks such as Tkinter, PyQt, or Kivy, using multithreading allows the main thread (which handles the user interface) to remain responsive while background tasks (like file uploads or complex calculations) are being executed in separate threads.

Without multithreading, the user interface might freeze or become unresponsive when performing lengthy tasks.

#  Parallelizing Independent Tasks
Multithreading is useful when you have independent tasks that can be executed in parallel, like processing multiple files, sending requests to a server, or handling multiple user interactions at once. It divides the workload across multiple threads, allowing tasks to be completed more efficiently.

```python
# Q.15 Write a Python program that prints the content of a file and handles the case when the file is empty

def print_file_content(filename):
    try:
        with open(filename, 'r') as file:
            content = file.read()
            if content:
                print("File Content:\n")
                print(content)
            else:
                print("The file is empty.")
    except FileNotFoundError:
        print(f"Error: The file '{filename}' does not exist.")
    except IOError as e:
        print(f"An I/O error occurred: {e}")

# Example usage:
filename = input("Enter the filename: ")
print_file_content(filename)
```

```python
# Q.18  How would you implement a basic logging setup that logs to a file with rotation after 1MB ?

import logging
import logging.handlers

# Create a logger
logger = logging.getLogger('my_logger')
logger.setLevel(logging.DEBUG)  # Set the desired logging level

# Create a RotatingFileHandler
handler = logging.handlers.RotatingFileHandler(
    'my_log.log',  # Log file name
    maxBytes=1024 * 1024,  # 1MB (1024 bytes * 1024)
    backupCount=5  # Keep 5 backup log files
)

# Set the formatter for the handler
formatter = logging.Formatter('%(asctime)s - %(name)s - %(levelname)s - %(message)s')
handler.setFormatter(formatter)

# Add the handler to the logger
logger.addHandler(handler)

# Now you can use the logger to log messages
logger.debug('This is a debug message.')
logger.info('This is an info message.')
logger.warning('This is a warning message.')
logger.error('This is an error message.')
logger.critical('This is a critical message.')
```

```python
# Q.21  Write a Python program that reads a file and prints the number of occurrences of a specific word.

def count_word_occurrences(filename, word):
    """
    Reads a file and counts the occurrences of a specific word.

    Args:
        filename: The name of the file to read.
        word: The word to count.

    Returns:
        The number of occurrences of the word in the file.
    """
    count = 0
    try:
        with open(filename, 'r') as file:
            for line in file:
                # Split the line into words and count occurrences
                count += line.lower().split().count(word.lower())
    except FileNotFoundError:
        print(f"Error: File '{filename}' not found.")
    except IOError as e:
        print(f"An I/O error occurred: {e}")
    return count

# Get the filename and word from the user
filename = input("Enter the filename: ")
word = input("Enter the word to count: ")

# Count the occurrences of the word
occurrences = count_word_occurrences(filename, word)

# Print the result
print(f"The word '{word}' appears {occurrences} times in the file.")
```
